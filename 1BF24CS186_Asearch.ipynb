```python
import heapq

# Define the goal state for a standard 3x3 8-puzzle (0 represents the empty blank tile)
GOAL_STATE = (1, 2, 3, 
              8, 5, 6, 
              7, 8, 0)

def get_misplaced_tiles(state, goal=GOAL_STATE):
    """
    Heuristic function h(n): Counts how many numbered tiles 
    are not in their correct goal positions (ignores the blank tile 0).
    """
    count = 0
    for i in range(len(state)):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count

def get_neighbors(state):
    """
    Generates all valid next states by shifting tiles 
    into the empty space (0) Up, Down, Left, or Right.
    """
    neighbors = []
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3
    
    # Define possible moves as (row_offset, col_offset, move_direction_name)
    moves = [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]
    
    for dr, dc, move_name in moves:
        new_row, new_col = row + dr, col + dc
        
        # Check boundary conditions
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank_idx = new_row * 3 + new_col
            
            # Convert tuple to list to swap items, then convert back
            new_state = list(state)
            new_state[blank_idx], new_state[new_blank_idx] = new_state[new_blank_idx], new_state[blank_idx]
            neighbors.append((tuple(new_state), move_name))
            
    return neighbors

def print_board(state):
    """Prints the state formatted neatly as a 3x3 grid."""
    for i in range(0, 9, 3):
        print(f"[{state[i]} {state[i+1]} {state[i+2]}]")
    print()

def solve_a_star(start_state):
    """
    Main A* Search algorithm loop.
    f(n) = g(n) + h(n)
    """
    # Priority Queue holds elements format: (f_score, g_score, current_state, path_taken)
    # g_score represents depth (number of moves made from start)
    start_h = get_misplaced_tiles(start_state)
    open_list = [(start_h, 0, start_state, [])]
    
    # Visited set tracks minimum g_score seen for a state to prevent redundant loops
    visited = {start_state: 0}
    
    nodes_expanded = 0
    
    while open_list:
        f, g, current, path = heapq.heappop(open_list)
        nodes_expanded += 1
        
        # Goal check
        if current == GOAL_STATE:
            return path, nodes_expanded
            
        # Explore neighbors
        for neighbor, move in get_neighbors(current):
            new_g = g + 1
            
            # If the neighbor hasn't been visited or we found a shorter path to it
            if neighbor not in visited or new_g < visited[neighbor]:
                visited[neighbor] = new_g
                new_h = get_misplaced_tiles(neighbor)
                new_f = new_g + new_h
                heapq.heappush(open_list, (new_f, new_g, neighbor, path + [(move, neighbor)]))
                
    return None, nodes_expanded

# ==========================================
# TEST IMPLEMENTATION IN YOUR JUPYTER CELL
# ==========================================

# 0 represents the blank space
initial_state = (2, 8, 3, 
                 1, 6, 4, 
                 7, 0, 5)

print("--- INITIAL BOARD STATE ---")
print_board(initial_state)

solution_path, total_nodes = solve_a_star(initial_state)

if solution_path is not None:
    print(f"🎉 Success! Goal reached in {len(solution_path)} moves.")
    print(f"Total nodes expanded: {total_nodes}\n")
    
    # Trace steps
    current_step = initial_state
    for step_num, (move, next_state) in enumerate(solution_path, 1):
        print(f"Step {step_num}: Move Blank Tile '{move}'")
        print_board(next_state)
else:
    print("❌ No solution found.")
```
