```python
# Objective/Heuristic function: returns the number of attacking pairs of queens
def get_cost(state):
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check if queens share the same row or the same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost
```

```python
# Generate all neighbors by moving each queen to a different row in its column
def get_neighbors(state):
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(n):
            if state[i] != j:
                neighbor = list(state)
                neighbor[i] = j
                neighbors.append(neighbor)
    return neighbors
```

```python
def hill_climbing(initial_state):
    current = list(initial_state)
    current_cost = get_cost(current)
    print(f"Your Initial State: {current} with cost (attacking pairs): {current_cost}")
    
    step = 0
    while True:
        neighbors = get_neighbors(current)
        best_neighbor = current
        best_cost = current_cost
        
        # Find the best neighbor with the minimum cost
        for neighbor in neighbors:
            temp_cost = get_cost(neighbor)
            if temp_cost < best_cost:
                best_cost = temp_cost
                best_neighbor = neighbor
                
        # If no neighbor is better than the current one, stop (reached a peak/local optimum)
        if best_cost >= current_cost:
            break
            
        current = best_neighbor
        current_cost = best_cost
        step += 1
        print(f"Step {step}: {current} with cost: {current_cost}")
        
        if current_cost == 0:
            break
            
    return current, current_cost
```
